```python
def Dayly_spreads(path_in,tiker,path_out):
    df = pd.read_csv(path_in+'/'+tiker+'.csv').drop(['Unnamed: 0','<PER>'], axis = 1)
    df_spreads = pd.DataFrame(columns = ['<TICKER>','<DATE>','<LAST>','<SPREAD>','<VOL>'])
    
    df['<DATE>']=df['<DATE>'].apply(lambda x: dt.datetime.strptime(x, "%d/%m/%y").date())
    df=df.sort_values('<DATE>')
    df['<DATE>']=df['<DATE>'].apply(lambda x: x.strftime('%Y-%m-%d'))
    df = df.sort_values(['<DATE>','<TIME>'])
    
    Delta = pd.DataFrame()
    Delta = df[['<DATE>','<TIME>','<LAST>']]
    Delta = Delta.sort_values(['<DATE>','<TIME>'])
    Delta['<SPREAD>'] = 0
    Delta['<SPREAD>'] = abs(Delta['<LAST>'] - Delta['<LAST>'].shift(1))
    df = df.join(Delta['<SPREAD>'])
    Delta = pd.DataFrame()
    
    df_spreads = df.groupby(by = '<DATE>')[['<TICKER>']].agg('max')
    last = df.groupby(by = '<DATE>')[['<DATE>','<LAST>']].tail(1).set_index('<DATE>')
    vol = df.groupby(by = '<DATE>')[['<VOL>']].agg('sum')
    date = df.groupby(by = '<DATE>')[['<DATE>']].agg('min')
    spread = df.groupby(by = '<DATE>')[['<SPREAD>']].agg('mean')
    
    df_spreads = df_spreads.join([date,last,spread,vol])
    
    df = pd.DataFrame()
    last = pd.DataFrame()
    vol = pd.DataFrame()
    date = pd.DataFrame()
    spread = pd.DataFrame()
    
    index = pd.Index(range(0,len(df_spreads)))
    df_spreads = df_spreads.set_index(index)
    
    df_spreads['<DATE>']=df_spreads['<DATE>'].apply(lambda x: dt.datetime.strptime(x, "%Y-%m-%d").date())
    df_spreads=df_spreads.sort_values('<DATE>')
    df_spreads['<DATE>']=df_spreads['<DATE>'].apply(lambda x: x.strftime('%d-%m-%Y'))

    df_spreads.to_csv(path_out+'/'+tiker+'_spreads.csv')
    return(df_spreads)
```
